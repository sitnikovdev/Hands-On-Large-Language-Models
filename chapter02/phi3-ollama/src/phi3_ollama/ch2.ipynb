```python
14. Ручная авторегрессия — обходной путь

Настоящего доступа к model(input_ids) нет, но можно сымитировать цикл, 
генерируя по одному токену за раз и каждый раз дописывая промпт (используя prompt caching, 
который мы видели в логе — это даже эффективно, так как Ollama переиспользует KV-кэш совпадающего префикса):
```

```python
text = prompt
for _ in range(700):
    resp = ollama.generate(model='phi3:mini', prompt=text, raw=True,
                            options={'temperature': 0, 'num_predict': 1})
    next_piece = resp['response']
    print(next_piece, end='', flush=True)
    text += next_piece
```

```python
16. Prompt caching — измеряем эффект вживую

Мы видели в логе, что повторный похожий промпт обработался почти мгновенно (f_sim_best = 1.000). Замерим это сами:
```

```python
import time

prompt = "Write an email apologizing to Sarah for the tragic gardening mishap.<|assistant|>"

# Первый вызов — холодный кэш
t0 = time.time()
resp1 = ollama.generate(model='phi3:mini', prompt=prompt, raw=True, options={'temperature': 0, 'num_predict': 20})
print(f"1-й вызов: {time.time()-t0:.2f}с, prompt_eval_count={resp1['prompt_eval_count']}")

# Второй вызов — тот же промпт, кэш префикса должен сработать
t0 = time.time()
resp2 = ollama.generate(model='phi3:mini', prompt=prompt, raw=True, options={'temperature': 0, 'num_predict': 20})
print(f"2-й вызов: {time.time()-t0:.2f}с, prompt_eval_count={resp2['prompt_eval_count']}")

# Совсем другой промпт — кэш не поможет
t0 = time.time()
resp3 = ollama.generate(model='phi3:mini', prompt="Explain quantum computing<|assistant|>", raw=True, options={'temperature': 0, 'num_predict': 20})
print(f"Другой промпт: {time.time()-t0:.2f}с, prompt_eval_count={resp3['prompt_eval_count']}")
```

```python
Хорошая иллюстрация того, почему поле n_keep = 4 и LCP-similarity в логе вообще существуют — это не абстракция, а реальная экономия времени.
```
